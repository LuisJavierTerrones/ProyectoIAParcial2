# LogiSmart AI - Ejecucion en Google Colab
Este cuaderno ejecuta el proyecto integrado. Primero sube `LogiSmart_AI.zip` desde tu computadora.

```python
from google.colab import files
uploaded = files.upload()  # Selecciona LogiSmart_AI.zip
```

```python
import zipfile, os
zip_name = next(name for name in uploaded if name.lower().endswith('.zip'))
with zipfile.ZipFile(zip_name) as z:
    z.extractall('/content')
os.chdir('/content/LogiSmart_AI')
print('Carpeta actual:', os.getcwd())
```
